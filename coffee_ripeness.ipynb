```python
import os
import shutil
import pandas as pd
from IPython.display import Image, display
from ultralytics import YOLO

# 1. Khôi phục dataset từ Drive vào /content/coffee_dataset
src_dataset = "/content/drive/MyDrive/Colab Notebooks/Coffee Cherry Ripeness.v1-2025-02-05-2-16pm.yolov11"
dst_dataset = "/content/coffee_dataset"

if not os.path.exists(dst_dataset):
    print("Đang sao chép dataset từ Drive sang Colab...")
    shutil.copytree(src_dataset, dst_dataset)
    print("-> Đã sao chép xong!")

# 2. Đánh giá mô hình trên tập test và lưu kết quả
weights_path = "/content/drive/MyDrive/coffee_runs/yolo11n_coffee/weights/best.pt"
model = YOLO(weights_path)

metrics = model.val(
    data=os.path.join(dst_dataset, "data.yaml"),
    split="test",
    project="/content/coffee_eval",
    name="test_results",
    exist_ok=True
)

# 3. Hiển thị ảnh Confusion Matrix
save_dir = metrics.save_dir
print("\n" + "="*50)
print("1. MA TRẬN NHẦM LẪN (CONFUSION MATRIX)")
print("="*50)

norm_path = os.path.join(save_dir, "confusion_matrix_normalized.png")
if os.path.exists(norm_path):
    display(Image(norm_path, width=700))
else:
    print("Chưa tìm thấy ảnh ma trận:", norm_path)

# 4. Trích xuất Classification Report
print("\n" + "="*50)
print("2. BÁO CÁO PHÂN LOẠI (CLASSIFICATION REPORT)")
print("="*50)

class_names = list(metrics.names.values())
p = metrics.box.p
r = metrics.box.r
f1 = metrics.box.f1
ap50 = metrics.box.ap50
ap = metrics.box.ap

df_classes = pd.DataFrame({
    'Lớp (Class)': class_names,
    'Precision': [round(x, 3) for x in p],
    'Recall': [round(x, 3) for x in r],
    'F1-Score': [round(x, 3) for x in f1],
    'mAP@50': [round(x, 3) for x in ap50],
    'mAP@50-95': [round(x, 3) for x in ap]
})

all_f1 = (2 * metrics.box.mp * metrics.box.mr) / (metrics.box.mp + metrics.box.mr + 1e-6)
df_all = pd.DataFrame([{
    'Lớp (Class)': 'Tất cả (all)',
    'Precision': round(metrics.box.mp, 3),
    'Recall': round(metrics.box.mr, 3),
    'F1-Score': round(all_f1, 3),
    'mAP@50': round(metrics.box.map50, 3),
    'mAP@50-95': round(metrics.box.map, 3)
}])

final_report = pd.concat([df_all, df_classes], ignore_index=True)
display(final_report)
```

```python
import os
import pandas as pd
from IPython.display import Image, display
from ultralytics import YOLO

# 1. Tải trọng số tốt nhất vừa train xong
best_weights = (
    "/content/drive/MyDrive/coffee_runs/yolov8n_coffee_optimized/weights/best.pt"
)
model_yolov8 = YOLO(best_weights)

# 2. Đánh giá trên tập test (lưu kết quả vào thư mục riêng)
metrics_yolov8 = model_yolov8.val(
    data="/content/coffee_dataset/data.yaml",
    split="test",
    imgsz=800,
    project="/content/coffee_eval",
    name="test_results_yolov8n",
    exist_ok=True,
)

# 3. Hiển thị Confusion Matrix (Chuẩn hóa)
print("\n" + "=" * 55)
print("1. CONFUSION MATRIX (YOLOv8n - Test Set)")
print("=" * 55)

conf_matrix_path = os.path.join(
    metrics_yolov8.save_dir, "confusion_matrix_normalized.png"
)
if os.path.exists(conf_matrix_path):
  display(Image(conf_matrix_path, width=700))
else:
  fallback_path = os.path.join(metrics_yolov8.save_dir, "confusion_matrix.png")
  if os.path.exists(fallback_path):
    display(Image(fallback_path, width=700))
  else:
    print(f"Chưa tìm thấy ảnh ma trận tại: {conf_matrix_path}")

# 4. In bảng Báo cáo phân loại chi tiết giống YOLO11s
print("\n" + "=" * 55)
print("2. BÁO CÁO PHÂN LOẠI CHI TIẾT (YOLOv8n)")
print("=" * 55)

class_names_yolov8 = list(metrics_yolov8.names.values())
p_yolov8 = metrics_yolov8.box.p
r_yolov8 = metrics_yolov8.box.r
f1_yolov8 = metrics_yolov8.box.f1
ap50_yolov8 = metrics_yolov8.box.ap50
ap_yolov8 = metrics_yolov8.box.ap

df_classes_yolov8 = pd.DataFrame({
    "Lớp quả": class_names_yolov8,
    "Precision": [round(x, 3) for x in p_yolov8],
    "Recall": [round(x, 3) for x in r_yolov8],
    "F1-Score": [round(x, 3) for x in f1_yolov8],
    "mAP@50": [round(x, 3) for x in ap50_yolov8],
    "mAP@50-95": [round(x, 3) for x in ap_yolov8],
})

all_p = metrics_yolov8.box.mp
all_r = metrics_yolov8.box.mr
all_f1 = (2 * all_p * all_r) / (all_p + all_r + 1e-6)

df_all_yolov8 = pd.DataFrame([{
    "Lớp quả": "Tất cả (all)",
    "Precision": round(all_p, 3),
    "Recall": round(all_r, 3),
    "F1-Score": round(all_f1, 3),
    "mAP@50": round(metrics_yolov8.box.map50, 3),
    "mAP@50-95": round(metrics_yolov8.box.map, 3),
}])

final_report_yolov8 = pd.concat(
    [df_all_yolov8, df_classes_yolov8], ignore_index=True
)
display(final_report_yolov8)
```
